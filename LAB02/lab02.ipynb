# Task 2

```python
A = np.array([
    [33, 22],
    [36, 44]
])

B = np.array([
    [56, 66],
    [71, 89]
])

print("Matrix A:")
print(A)

print("\nMatrix B:")
print(B)
```

```python
addition = A + B

print("A + B:")
print(addition)
```

```python
elementwise_multiplication = A * B

print("Element-wise multiplication (A * B):")
print(elementwise_multiplication)
```

```python
matrix_multiplication = A @ B

print("Matrix multiplication (A @ B):")
print(matrix_multiplication)
```

```python
transpose_A = A.T

print("Transpose of A:")
print(transpose_A)
```

```python
determinant_A = np.linalg.det(A)

print("Determinant of A:", determinant_A)
```

```python
inverse_A = np.linalg.inv(A)

print("Inverse of A:")
print(inverse_A)
```

```python
verification = A @ inverse_A

print("A @ A^-1:")
print(verification)

print("\nIs A @ A^-1 approximately equal to the identity matrix?")
print(np.allclose(verification, np.eye(2)))
```

```python
np.random.seed(42)  # For reproducibility

CHOSEN_MEAN = 35
CHOSEN_STD = 18
N_SAMPLES = 1000

normal_data = np.random.normal(loc=CHOSEN_MEAN, scale=CHOSEN_STD, size=N_SAMPLES)

sample_mean = normal_data.mean()
sample_std = normal_data.std()

print(f"Chosen (population) mean : {CHOSEN_MEAN}")
print(f"Chosen (population) std  : {CHOSEN_STD}")
print(f"Sample mean              : {sample_mean:.4f}")
print(f"Sample std               : {sample_std:.4f}")
print(f"Difference in mean       : {abs(sample_mean - CHOSEN_MEAN):.4f}")
print(f"Difference in std        : {abs(sample_std - CHOSEN_STD):.4f}")
print("\nThe sample statistics are very close to the chosen population parameters,"
      " which is expected with n=1000 draws (law of large numbers).")
```

```python
import matplotlib.pyplot as plt

plt.hist(normal_data, bins=30, edgecolor="black")

plt.xlabel("Value")
plt.ylabel("Frequency")
plt.title("Histogram of Normally Distributed Data")

plt.show()
```

```python
missing_count = df.isnull().sum()
missing_percentage = (
    df.isnull().sum() / len(df) * 100
).round(2)
missing_summary = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

print("Missing values by column:")
display(missing_summary)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))

missing_count.plot(kind="bar")

plt.title("Missing Values by Column")
plt.xlabel("Column")
plt.ylabel("Missing Count")
plt.xticks(rotation=45)

plt.tight_layout()

# Save directly in the current folder
plt.savefig("task7_missing_values_bar.png", dpi=120)

plt.show()
```

```python
age_missing_before = df["Age"].isnull().sum()

print("Missing Age values before:", age_missing_before)
mean_age = df["Age"].mean()

print("Mean Age:", round(mean_age, 2))
df_mean = df.copy()
df_mean["Age"] = df_mean["Age"].fillna(mean_age)
age_missing_after = df_mean["Age"].isnull().sum()

print("Missing Age values after:", age_missing_after)
```

```python
import numpy as np
original_missing = df["Age"].isnull().sum()

print("Original missing Age values:", original_missing)
df_mean = df.copy()

mean_age = df_mean["Age"].mean()

df_mean["Age"] = df_mean["Age"].fillna(mean_age)

print("\n1. Mean Imputation")
print("Mean Age:", round(mean_age, 2))
print("Missing Age after:", df_mean["Age"].isnull().sum())

df_median = df.copy()

median_age = df_median["Age"].median()

df_median["Age"] = df_median["Age"].fillna(median_age)

print("\n2. Median Imputation")
print("Median Age:", round(median_age, 2))
print("Missing Age after:", df_median["Age"].isnull().sum())

df_mode = df.copy()

mode_age = df_mode["Age"].mode()[0]

df_mode["Age"] = df_mode["Age"].fillna(mode_age)

print("\n3. Mode Imputation")
print("Mode Age:", mode_age)
print("Missing Age after:", df_mode["Age"].isnull().sum())

df_random = df.copy()
np.random.seed(42)
available_ages = df_random["Age"].dropna()
random_ages = np.random.choice(
    available_ages,
    size=df_random["Age"].isnull().sum(),
    replace=True
)
df_random.loc[
    df_random["Age"].isnull(),
    "Age"
] = random_ages

print("\n4. Random Value Imputation")
print("Missing Age after:", df_random["Age"].isnull().sum())
```

```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
fare_outliers = df[
    (df["Fare"] < lower_bound) |
    (df["Fare"] > upper_bound)
]
outlier_count = len(fare_outliers)

print("----- FARE OUTLIER ANALYSIS -----")

print("Q1:", round(Q1, 2))
print("Q3:", round(Q3, 2))
print("IQR:", round(IQR, 2))

print("Lower Bound:", round(lower_bound, 2))
print("Upper Bound:", round(upper_bound, 2))

print("Number of Fare Outliers:", outlier_count)
```
